```python
from nba_api.stats.endpoints import playercareerstats
from nba_api.stats.static import players
from nba_api.stats.static import teams
from nba_api.stats.endpoints import commonteamroster
```

```python
# Get all NBA teams
nba_teams = teams.get_teams()

# Display all team names
print("All NBA Team Names:")
print("-" * 30)
for team in nba_teams:
    print(f"{team['full_name']} ({team['abbreviation']})")
```

```python
# Nikola Jokić
career = playercareerstats.PlayerCareerStats(player_id='203999') 

career.get_data_frames()[0]
```

```python
nba_players = players.get_players()

print("\nAll NBA Players:")
print("-" * 30)
for player in nba_players:
    print(f"{player['full_name']} ({player['id']})")
```

```python
number_of_players = len(nba_players)
print("Total number of NBA players ever:", number_of_players)
```

```python
clippers = teams.find_teams_by_full_name('Los Angeles Clippers')[0]
clippers

clippers_roster = commonteamroster.CommonTeamRoster(team_id=clippers['id'])
clippers_players = clippers_roster.get_data_frames()[0]
clippers_players.head()

for player in clippers_players['PLAYER']:
    print(player)
```
