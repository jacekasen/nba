```python
from pathlib import Path

import pandas as pd

DATA_PATH = Path("..") / "data" / "nba_player_stats.csv"

df = pd.read_csv(DATA_PATH)

print(f"Shape: {df.shape[0]:,} rows × {df.shape[1]} columns")
print(f"Seasons: {df['year_id'].nunique()} ({df['year_id'].min()} → {df['year_id'].max()})")
print(f"Players: {df['player_name'].nunique():,}")
print(f"Teams:   {df['team_name_abbr'].nunique()}")
print()

display(df.dtypes.to_frame("dtype"))
display(df.isna().sum().to_frame("nulls").assign(pct=lambda x: (x["nulls"] / len(df) * 100).round(1)))
display(df.describe())
display(df.head(10))
```

```python
# Basketball-Reference already provides a season total row (2TM/3TM/…).
# Keep that aggregate and drop the per-team split rows for the same player-season.
ntm_pattern = r"^\d+TM$"
is_ntm = df["team_name_abbr"].str.match(ntm_pattern, na=False)
season_has_ntm = df.groupby(["player_url", "year_id"])["team_name_abbr"].transform(
    lambda s: s.str.match(ntm_pattern, na=False).any()
)

before = len(df)
df = df.loc[~season_has_ntm | is_ntm].copy()

n_ntm = df["team_name_abbr"].str.match(ntm_pattern, na=False).sum()
rows_per_season = df.groupby(["player_url", "year_id"]).size()
n_multi = (rows_per_season > 1).sum()

print(f"Merged multi-team seasons: dropped {before - len(df):,} team-split rows ({before:,} → {len(df):,})")
print(f"Remaining nTM total rows: {n_ntm:,}")
print(f"Player-seasons with >1 row left: {n_multi}")
```

```python
# Exact scrape duplicates (e.g. Danté Exum) + alternate-name duplicates
# (Jeenathan Williams / Nate Williams share willije02).
before = len(df)

# Modern Nate Williams is Jeenathan; keep the BRef name for that URL.
jeenathan_url = "https://www.basketball-reference.com/players/w/willije02.html"
df.loc[df["player_url"] == jeenathan_url, "player_name"] = "Jeenathan Williams"

df = df.drop_duplicates()
df = df.drop_duplicates(subset=["player_url", "year_id"], keep="first").copy()

rows_per_season = df.groupby(["player_url", "year_id"]).size()
print(f"Dropped {before - len(df):,} duplicate player-season rows ({before:,} → {len(df):,})")
print(f"Player-seasons with >1 row left: {(rows_per_season > 1).sum()}")
```

```python
# ABA existed 1967-68 → 1975-76; four teams joined the NBA for 1976-77.
# Drop ABA team seasons before the merger (keep contemporaneous NBA seasons).
ABA_TEAMS = {
    "ANA", "AND", "CAR", "DLC", "DNA", "DNR", "FLO", "HSM", "INA", "KEN", "LAS",
    "MMF", "MMS", "MMT", "MNM", "NJA", "NOB", "NYA", "OAK", "PTP", "PTC", "SAA",
    "SDA", "SDS", "SSL", "TEX", "UTS", "VIR",
}
MERGER_SEASON = "1976-77"

# nTM totals that include any ABA club (incl. rare NBA↔ABA swaps)
raw_pre = pd.read_csv(DATA_PATH)
raw_pre["player_url"] = raw_pre["player_url"].fillna(raw_pre["player_name"].map(missing_urls))
raw_pre = raw_pre.loc[
    (raw_pre["year_id"] < MERGER_SEASON)
    & ~raw_pre["team_name_abbr"].str.startswith("Did not play", na=False)
    & ~raw_pre["team_name_abbr"].str.match(r"^\d+TM$", na=False)
]
aba_season_keys = set(
    zip(
        raw_pre.loc[raw_pre["team_name_abbr"].isin(ABA_TEAMS), "player_url"],
        raw_pre.loc[raw_pre["team_name_abbr"].isin(ABA_TEAMS), "year_id"],
    )
)

pre_merger = df["year_id"] < MERGER_SEASON
is_aba_team = df["team_name_abbr"].isin(ABA_TEAMS)
is_ntm = df["team_name_abbr"].str.match(r"^\d+TM$", na=False)
ntm_touches_aba = [
    (url, year) in aba_season_keys
    for url, year in zip(df["player_url"], df["year_id"])
]
drop_aba = pre_merger & (is_aba_team | (is_ntm & pd.Series(ntm_touches_aba, index=df.index)))

before = len(df)
df = df.loc[~drop_aba].copy()

print(f"Dropped {before - len(df):,} pre-merger ABA rows ({before:,} → {len(df):,})")
print(f"Seasons remaining: {df['year_id'].min()} → {df['year_id'].max()}")
print(f"Pre-merger rows left (NBA only): {(df['year_id'] < MERGER_SEASON).sum():,}")
print(f"ABA team codes left: {df['team_name_abbr'].isin(ABA_TEAMS).sum()}")
```

```python
# Additional hardening for an analysis-ready dataset
# Run this notebook inside the conda environment: nba-peak-analysis

MERGER_SEASON = "1976-77"
numeric_cols = ["age", "games", "mp", "per", "bpm", "vorp", "ws", "ws_per_48"]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors="coerce")

before = len(df)
df = df.loc[~(df["year_id"] < MERGER_SEASON)].copy()

# Drop rows that are still incomplete or nonsensical for analysis
required_cols = ["player_name", "player_url", "team_name_abbr", "age", "games", "mp"]
df = df.loc[df[required_cols].notna().all(axis=1)].copy()
df = df.loc[(df["games"] > 0) & (df["mp"] > 0) & (df["age"].notna())].copy()

# Keep a stable row order for downstream analysis
# (player-season keys should be unique after earlier de-duplication)
df = df.sort_values(["player_url", "year_id"]).reset_index(drop=True)

print(f"Dropped {before - len(df):,} rows after removing pre-merger seasons and invalid records ({before:,} → {len(df):,})")
print(f"Seasons remaining: {df['year_id'].min()} → {df['year_id'].max()}")
print(f"Rows with remaining nulls:\n{df[numeric_cols + required_cols].isna().sum()}")

# Optional: save the stricter cleaned dataset for later notebooks
out_path = Path("..") / "data" / "03-cleaned" / "nba_complete_clean.csv"
out_path.parent.mkdir(parents=True, exist_ok=True)
df.to_csv(out_path, index=False)
print(f"Saved stricter cleaned dataset to: {out_path}")
```

```python
# --- Anomaly / completeness checks on cleaned df ---
metric_cols = ["per", "bpm", "vorp", "ws", "ws_per_48"]
id_cols = ["player_url", "year_id"]

qa = df.copy()
qa["season_start"] = qa["year_id"].str.slice(0, 4).astype(int)

print("=== Coverage ===")
print(f"Rows: {len(qa):,} | players: {qa['player_url'].nunique():,} | seasons: {qa['year_id'].nunique()} ({qa['year_id'].min()} → {qa['year_id'].max()})")
print(f"Nulls in key cols:\n{qa[id_cols + ['player_name', 'age', 'games', 'mp'] + metric_cols].isna().sum()}")

print("\n=== Identity / uniqueness ===")
dup_seasons = qa.duplicated(subset=id_cols, keep=False)
print(f"Duplicate player-season keys: {dup_seasons.sum()}")
names_per_url = qa.groupby("player_url")["player_name"].nunique()
urls_per_name = qa.groupby("player_name")["player_url"].nunique()
print(f"URLs with >1 display name: {(names_per_url > 1).sum()}")
print(f"Display names with >1 URL (homonyms): {(urls_per_name > 1).sum()}")
if (urls_per_name > 1).any():
    display(urls_per_name[urls_per_name > 1].sort_values(ascending=False).head(15).to_frame("n_urls"))

print("\n=== Career season gaps (missing years between first and last) ===")
# Gaps are often real (injury / overseas / G-League); flag for review.
gap_rows = []
for url, g in qa.groupby("player_url"):
    years = sorted(g["season_start"].unique())
    name = g["player_name"].iloc[0]
    for prev, curr in zip(years, years[1:]):
        hole = curr - prev - 1
        if hole > 0:
            gap_rows.append(
                {
                    "player_name": name,
                    "player_url": url,
                    "after_season": prev,
                    "before_season": curr,
                    "missing_years": hole,
                    "career_seasons": len(years),
                }
            )
gaps = pd.DataFrame(gap_rows)
print(f"Players with ≥1 gap: {gaps['player_url'].nunique() if len(gaps) else 0}")
print(f"Total gap segments: {len(gaps)}")
if len(gaps):
    display(gaps.sort_values("missing_years", ascending=False).head(15))

print("\n=== Age consistency across seasons ===")
qa = qa.sort_values(["player_url", "season_start"])
qa["age_delta"] = qa.groupby("player_url")["age"].diff()
qa["year_delta"] = qa.groupby("player_url")["season_start"].diff()
age_bad = qa[qa["year_delta"].notna() & ((qa["age_delta"] - qa["year_delta"]).abs() > 1)]
print(f"Seasons where age jump ≠ year jump (±1 tolerance): {len(age_bad)}")
if len(age_bad):
    display(age_bad[["player_name", "year_id", "age", "age_delta", "year_delta"]].head(20))

print("\n=== Playing-time sanity ===")
print(f"Rows with mp < games: {(qa['mp'] < qa['games']).sum()}")
print(f"Rows with games > 82: {(qa['games'] > 82).sum()} (ABA / 3-team seasons can exceed 82)")
print(f"Rows with games == 0: {(qa['games'] == 0).sum()}")

print("\n=== Completeness vs raw (same metric filters, before multi-team collapse) ===")
raw = pd.read_csv(DATA_PATH)
raw["player_url"] = raw["player_url"].fillna(raw["player_name"].map(missing_urls))
raw = raw.loc[~raw["team_name_abbr"].str.startswith("Did not play", na=False)].copy()
raw = raw.dropna(subset=["age"] + metric_cols)
# expected unique player-seasons if we keep nTM totals only
ntm = raw["team_name_abbr"].str.match(r"^\d+TM$", na=False)
raw["_has_ntm"] = raw.groupby(["player_url", "year_id"])["team_name_abbr"].transform(
    lambda s: s.str.match(r"^\d+TM$", na=False).any()
)
raw_expected = raw.loc[~raw["_has_ntm"] | ntm, id_cols].drop_duplicates()
raw_expected = raw_expected.drop_duplicates(subset=id_cols)

clean_keys = qa[id_cols].drop_duplicates()
missing = raw_expected.merge(clean_keys, on=id_cols, how="left", indicator=True)
missing = missing.loc[missing["_merge"] == "left_only", id_cols]
extra = clean_keys.merge(raw_expected, on=id_cols, how="left", indicator=True)
extra = extra.loc[extra["_merge"] == "left_only", id_cols]

print(f"Expected player-seasons (raw, metric-complete, nTM-collapsed): {len(raw_expected):,}")
print(f"Cleaned player-seasons: {len(clean_keys):,}")
print(f"Missing from cleaned: {len(missing)}")
print(f"Extra in cleaned: {len(extra)}")
if len(missing):
    display(missing.head(20))
```
